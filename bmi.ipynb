```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\jabes\OneDrive\Documents\Dataset\bmi.csv")
```

```python
print("Exploratory Data Analysis :")
display(df.head())
display(df.tail())
print(f" Total Baris : {df.shape[0]} Baris")
print(f" Total Kolom : {df.shape[1]} Kolom")
print(f" Total Nilai Nan : {df.isna().sum().sum()}")
print(f" Total Nilai Duplicated : {df.duplicated().sum()}")
```

```python
print("pengecekan nilai oulier :".title())
plt.figure(figsize=(12,6))
sns.boxplot(data= df, orient='h')
```

```python
print(" Deskripsi Kolom :")
df.describe().T
```

```python
print(" Data Cleaning")
df = df.drop_duplicates().reset_index(drop=True)
```

```python
print('pengecekan Nilai Korelasi :')
corr = df.corr(numeric_only=True)
plt.figure(figsize=(12,2))
sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)
```
