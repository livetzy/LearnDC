```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x = np.array([1, 2, 3, 4, 5,])
y = np.array([1, 4, 9 , 16 , 25])
y2 = np.array([1, 16, 81, 256, 625])
```

```python
plt.plot(x, y , 'red')
plt.plot(y2, 'blue')
plt.title("MY FIRST PLOT")
plt.legend([r'${line{1}}$', 'line2'])
plt.xlabel("X AXIS")
plt.ylabel("Y AXIS")
plt.show()
```

```python
def sinusGenerator(amplitudo, frekuensi, tAkhir, theta):
    t = np.arange(0, tAkhir, 0.1)
    y = amplitudo * np.sin(2 * frekuensi * t + np.deg2rad(theta))
    return t,y
      
t1 , y1 = sinusGenerator(1,1,4,0)
plt.plot(t1, y1)
```

```python
t2,y2 = sinusGenerator(1,1,4,30)
plt.plot(t2, y2, t1, y1, 'r')
plt.show()
```

```python
t3, y3 = sinusGenerator(1,1,4,30)
plt.plot(t3, y3, 'b-o', t2, y2, 'c--', t1, y1, 'r-o')
plt.show()
```

```python

t4, y4 = sinusGenerator(1,1,4,180)
t2, y2 = sinusGenerator(1,1,4,90)
t1, y1 = sinusGenerator(1,1,4,60)
plt.plot(t1, y1)
plt.plot(t2, y2)
plt.plot(t3, y3)
dataPlot4 = plt.plot(t4, y4)
dataPlot3 = plt.plot(t3, y3)
dataPlot2 = plt.plot(t2, y2)
dataPlot1 = plt.plot(t1, y1)
plt.setp(dataPlot4[0], color='r', linestyle='-.',linewidth=5)
plt.setp(dataPlot3[0], color='g', linestyle='--',linewidth=3)
plt.setp(dataPlot2[0], color='c', linestyle='--',linewidth=4)
plt.setp(dataPlot1[0], color='b', linestyle='--', linewidth=4)

judul = "Grafik Sinusoidal"
rumus = r"$ \mathcal{y} = A \sin(2 \pi f t + \theta)$"
plt.title(judul + "\n" + rumus)
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.legend(["60°", "90°", "180°", "0°"])
plt.grid(True)
plt.show()
```

```python
# Bagian 1: x < 0
x1 = np.linspace(-3, 0, 100, endpoint=False)
plt.plot(x1, x1 + 1, color="blue")

# Bagian 2: 0 <= x <= 2
x2 = np.linspace(0, 2, 100)
plt.plot(x2, x2**2, color="orange")

# Bagian 3: x > 2
x3 = np.linspace(2, 3, 100)[1:]  # tidak menyertakan x = 2
plt.plot(x3, np.full_like(x3, 2), color="green")

# Titik penuh berarti nilai tersebut termasuk dalam interval
plt.plot([0, 2], [0, 4], color='r')

# Lingkaran kosong berarti titik tersebut tidak termasuk
plt.plot([0, 2], [1, 2], "o", markerfacecolor="white", markeredgecolor="black")

plt.axhline(0, color="gray", linewidth=0.8)
plt.axvline(0, color="gray", linewidth=0.8)
plt.grid(True)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend(['f(x) = ax + c'])
plt.title('Fungsi Sepotong Sepotong')
plt.show()
```

```python
sudut = np.arange(0,360,1)
y = np.sin(np.deg2rad(sudut))
```

```python
plt.figure(figsize=(12, 5))
plt.plot(sudut,y , 'r--',)
plt.title('grafik sinusidal'.title())
plt.text(190,1, 'magnituda')
plt.text(360 , 0.1 , 'sudut')
plt.xticks([0, 90, 180, 270, 360],
           [r'${0}^o$', r'${90}^o$', r'${180}^o$', r'${270}^o$', r'${360}^o$',])
plt.yticks([-1, -0.5 , 0 , 0.5 ,1])



ax = plt.gca()
ax.spines['left'].set_position(('data', 180))
ax.spines['bottom'].set_position(('data', 0))
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')
plt.show()
```
