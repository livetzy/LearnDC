```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x = np.array([1, 2, 3, 4, 5,])
y = np.array([1, 4, 9 , 16 , 25])
y2 = np.array([1, 16, 81, 256, 625])
```

```python
plt.plot(x, y , y2)
plt.title("MY FIRST PLOT")
plt.xlabel("X AXIS")
plt.ylabel("Y AXIS")
plt.show()
```

```python
def sinusGenerator(amplitudo, frekuensi, tAkhir, theta):
    t = np.arange(0, tAkhir, 0.1)
    y = amplitudo * np.sin(2 * frekuensi * t + np.deg2rad(theta))
    return t,y
      
t1 , y1 = sinusGenerator(1,1,4,0)
plt.plot(t1, y1)
```

```python
t2,y2 = sinusGenerator(1,1,4,30)
plt.plot(t2, y2, t1, y1, 'r')
plt.show()
```

```python
t3, y3 = sinusGenerator(1,1,4,30)
plt.plot(t3, y3, 'b-o', t2, y2, 'c--', t1, y1, 'r-o')
plt.show()
```

```python

t4, y4 = sinusGenerator(1,1,4,180)
t2, y2 = sinusGenerator(1,1,4,90)
t1, y1 = sinusGenerator(1,1,4,60)
plt.plot(t1, y1)
plt.plot(t2, y2)
plt.plot(t3, y3)
dataPlot4 = plt.plot(t4, y4)
dataPlot3 = plt.plot(t3, y3)
dataPlot2 = plt.plot(t2, y2)
dataPlot1 = plt.plot(t1, y1)
plt.setp(dataPlot4[0], color='r', linestyle='-.',linewidth=5)
plt.setp(dataPlot3[0], color='g', linestyle='--',linewidth=3)
plt.setp(dataPlot2[0], color='c', linestyle='--',linewidth=4)
plt.setp(dataPlot1[0], color='b', linestyle='--', linewidth=4)

judul = "Grafik Sinusoidal"
rumus = r"$ \mathcal{y} = A \sin(2 \pi f t + \theta)$"
plt.title(judul + "\n" + rumus)
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.legend(["60°", "90°", "180°", "0°"])
plt.grid(True)
plt.show()
```
