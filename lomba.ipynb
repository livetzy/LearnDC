```python
print(" Data Cleaning :")
df["volume_l_missing"] = df["volume_l"].isna().astype("int8")

df["volume_l"] = df.groupby("package_id")["volume_l"].transform(
    lambda s: s.fillna(s.median())
)

# Untuk package_id yang median-nya juga tidak tersedia
df["volume_l"] = df["volume_l"].fillna(df["volume_l"].median())
```
